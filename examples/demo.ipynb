# DeepFMP-DG walkthrough
Load the packaged models and diagnose one item end-to-end.

```python
from deepfmp_dg.infer import InferenceEngine

engine = InferenceEngine()
print("InferenceEngine ready")
```

```python
from pathlib import Path

sellers = sorted(Path('data/sample/images/seller').glob('*.jpg'))
buyers = sorted(Path('data/sample/images/buyer').glob('*.jpg'))
assert sellers and buyers
result = engine.predict(
    seller_image=sellers[0],
    buyer_image=buyers[0],
    title='Sample product',
    description='Demo item for the walkthrough',
    review_text='Great quality, exactly as described.',
    price=19.99,
)
result
```

## Reading the output
`scores` are the three expectation-gap alignments; `risk_probability` comes from the fused network; `top_shap` explains which tabular features drive the prediction; `diagnosis` and `recommendations` are the triage output.
